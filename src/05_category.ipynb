```python
import pandas as pd
import numpy as np
```

```python
# 데이터 로드
df = pd.read_csv(r'../data/03_train.csv')

#카테고리 컬럼 추가

# 현재 Label 확인
print("\n=== Label Distribution ===")
print(df['Label'].value_counts())

# 카테고리 매핑 정의
category_mapping = {
    # DoS/DDoS 계열
    'DoS attacks-GoldenEye': 'DoS/DDoS',
    'DoS attacks-SlowHTTPTest': 'DoS/DDoS',
    'DDoS attacks-LOIC-HTTP': 'DoS/DDoS',
    'DDOS attack-LOIC-UDP': 'DoS/DDoS',
    
    # Brute Force (인증 서비스 대상)
    'FTP-BruteForce': 'Brute Force',
    
    # Web Attack (웹 서버 대상)
    'Brute Force -Web': 'Web Attack',
    'Brute Force -XSS': 'Web Attack',
    'SQL Injection': 'Web Attack',
    
    # Botnet
    'Bot': 'Botnet',
    
    # 정상
    'Benign': 'Benign'
}

# 카테고리 컬럼 추가
df['Category'] = df['Label'].map(category_mapping)

# 매핑 안 된 값 확인
unmapped = df[df['Category'].isna()]['Label'].unique()
if len(unmapped) > 0:
    print(f"\n매핑 안 된 Label: {unmapped}")
else:
    print("\n모든 Label 매핑 완료")

# 카테고리별 분포 확인
print("\n=== 카테고리별 분포 ===")
category_counts = df['Category'].value_counts()
print(category_counts)

print("\n=== 카테고리별 상세 ===")
for category in category_counts.index:
    print(f"\n[{category}] - {category_counts[category]} samples")
    labels_in_category = df[df['Category'] == category]['Label'].value_counts()
    for label, count in labels_in_category.items():
        print(f"  - {label}: {count} samples")

# 저장
output_path = r'../data/05_train_with_category.csv'
df.to_csv(output_path, index=False)

print(f"\n 저장 완료: {output_path}")
print(f"총 {len(df)}개 행, {len(df.columns)}개 컬럼 (Category 추가됨)")
```
